```python
import random
jackpot=random.randint(1,100)       # Generate the random number 
user_input=int(input ('guess the number'))
c=1
while jackpot != user_input:
    if user_input <jackpot:
        print('choose the higher')
        user_input=int(input('guess the number'))
        c+=1
    else:
        print('choose the lower')
        user_input=int(input('guess the number'))
        c+=1
else:
    print('you won the game')
    print('your attempt',c)
```

# nested loop

```python
# nested loop
for i in range(1,5):
    for j in range(1,5):
        print(i,j)
```

```python
# nested loop
num=int(input('enter the number'))
for i in range(1,num+1):
    for j in range (1,i+1):
        print( '#',end=' ')
    print()
```

```python
# nested loop
num=int(input('enter the number'))
for i in range(num,0,-1):
    for j in range (i):
        print( '#',end=' ')
    print()
```

```python
# nested loop
num=int(input('enter the number'))
for i in range(1,num+1):
    print( ' '*(num - i),end='')
    print('# '*i)
```
